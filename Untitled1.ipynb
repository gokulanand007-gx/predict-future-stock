```python
# ================= JSE STOCK CLOSING-PRICE PREDICTION (LSTM) =================
# Paste into ONE Colab cell. Upload JSEClosing.xlsx when prompted.
import numpy as np, pandas as pd, matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
from google.colab import files

np.random.seed(42); tf.random.set_seed(42)

TICKER = "SBKJ.J"     # change to any column: e.g. "MTNJ.J", "AGLJ.J", ".JTOPI" (index)
WINDOW = 60           # past 60 trading days -> next day

# ---------- 1. LOAD ----------
uploaded = files.upload()
fname = list(uploaded.keys())[0]

# The file is a pivot table: header on row 2, dates down rows, one column per stock
raw = pd.read_excel(fname, sheet_name="Sheet1", header=1).rename(columns={"Row Labels": "Date"})

# ---------- 2. CLEAN ----------
raw["Date"] = pd.to_datetime(raw["Date"], errors="coerce")        # drops "(blank)" / "Grand Total" rows
raw = raw.dropna(subset=["Date"]).drop_duplicates("Date").sort_values("Date").set_index("Date")
print("Available tickers:", [c for c in raw.columns if c not in ("Grand Total", "(blank)")])

close = pd.to_numeric(raw[TICKER], errors="coerce").dropna()      # blanks = weekends/holidays -> removed
close = close[close > 0]
print(f"{TICKER}: {len(close)} trading days, {close.index.min().date()} -> {close.index.max().date()}")

# ---------- 3. NORMALIZE + SEQUENCES ----------
# Model daily log-returns (stationary) instead of raw prices: prices trend beyond the
# training range, which breaks MinMax-scaled price models.
ret = np.log(close).diff().dropna()
n = len(ret)
train_end, val_end = int(n * 0.70), int(n * 0.85)

scaler = StandardScaler().fit(ret.iloc[:train_end].values.reshape(-1, 1))   # TRAIN only (no leakage)
z = scaler.transform(ret.values.reshape(-1, 1)).ravel()

tgt = np.arange(WINDOW, n)                                          # target positions
X = np.array([z[i - WINDOW:i] for i in tgt])[..., None]             # (samples, WINDOW, 1)
y = z[tgt]
prev_close = close.values[tgt]          # yesterday's close for each target
actual     = close.values[tgt + 1]      # the real close we want to predict
dates      = ret.index[tgt]

tr = tgt < train_end
va = (tgt >= train_end) & (tgt < val_end)
te = tgt >= val_end
print("Train:", X[tr].shape, "Val:", X[va].shape, "Test:", X[te].shape)

# ---------- 4. LSTM ----------
model = models.Sequential([
    layers.Input(shape=(WINDOW, 1)),
    layers.LSTM(64, return_sequences=True),
    layers.Dropout(0.2),
    layers.LSTM(32),
    layers.Dropout(0.2),
    layers.Dense(16, activation="relu"),
    layers.Dense(1),
])
model.compile(optimizer=tf.keras.optimizers.Adam(1e-3), loss="mse")
model.summary()

history = model.fit(
    X[tr], y[tr], validation_data=(X[va], y[va]),
    epochs=100, batch_size=32, shuffle=False, verbose=1,
    callbacks=[
        callbacks.EarlyStopping(monitor="val_loss", patience=12, restore_best_weights=True),
        callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=5, min_lr=1e-5),
    ],
)
plt.figure(figsize=(8, 4))
plt.plot(history.history["loss"], label="Train"); plt.plot(history.history["val_loss"], label="Val")
plt.title("Training curve"); plt.xlabel("Epoch"); plt.ylabel("MSE"); plt.legend(); plt.show()

# ---------- 5. EVALUATE (MAE, RMSE in price units) ----------
pred_ret = scaler.inverse_transform(model.predict(X[te]).reshape(-1, 1)).ravel()
y_pred   = prev_close[te] * np.exp(pred_ret)       # rebuild predicted price
y_actual = actual[te]
y_naive  = prev_close[te]                          # baseline: tomorrow = today

def report(name, a, p):
    mae, rmse = mean_absolute_error(a, p), np.sqrt(mean_squared_error(a, p))
    print(f"{name:6s} MAE: {mae:10.2f} | RMSE: {rmse:10.2f} | MAE as % of mean price: {100*mae/a.mean():.2f}%")
report("LSTM", y_actual, y_pred)
report("Naive", y_actual, y_naive)

# ---------- 6. PLOTS ----------
d_test = dates[te]
plt.figure(figsize=(12, 5))
plt.plot(d_test, y_actual, label="Actual Close", linewidth=2)
plt.plot(d_test, y_pred, label="Predicted Close", linewidth=1.3)
plt.title(f"{TICKER}: Actual vs Predicted Closing Price (unseen test period)")
plt.xlabel("Date"); plt.ylabel("Price"); plt.legend(); plt.grid(alpha=0.3); plt.show()

plt.figure(figsize=(12, 4))
plt.plot(d_test[-60:], y_actual[-60:], marker="o", label="Actual")
plt.plot(d_test[-60:], y_pred[-60:], marker="x", label="Predicted")
plt.title("Last 60 test days (zoomed)"); plt.legend(); plt.grid(alpha=0.3); plt.show()

# ---------- 7. UNSEEN-DATE PREDICTIONS ----------
demo = pd.DataFrame({"Date": d_test[-10:].date, "Actual": y_actual[-10:].round(2),
                     "Predicted": y_pred[-10:].round(2)})
demo["Abs_Error"] = (demo["Actual"] - demo["Predicted"]).abs().round(2)
print(demo.to_string(index=False))

last = z[-WINDOW:].reshape(1, WINDOW, 1)
next_ret = scaler.inverse_transform(model.predict(last))[0, 0]
next_price = close.values[-1] * np.exp(next_ret)
next_date = pd.bdate_range(close.index[-1], periods=2)[-1]
print(f"\nForecast for {next_date.date()}: {next_price:.2f}  (last close {close.values[-1]:.2f})")
```
